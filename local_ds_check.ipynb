```python
import numpy as np
import netCDF4 as nc

# 1. Open the file
ds = nc.Dataset('06.nc', 'r')
var = ds.variables['t2m']

# 2. Get chunk dimensions and data type
chunks = var.chunking()
dtype_size = var.dtype.itemsize  # Size in bytes

if chunks == 'contiguous':
    print("Variable is contiguous. The entire variable is one chunk.")
else:
    # 3. Calculate total values per chunk
    num_elements = np.prod(chunks)
    
    # 4. Calculate size in MB
    chunk_size_mb = (num_elements * dtype_size) / (1024 ** 2)
    
    print(f"Dimensions order: {var.dimensions}")
    print(f"Chunk shape: {chunks}")
    print(f"Data type: {var.dtype} ({dtype_size} bytes)")
    print(f"Uncompressed size per chunk: {chunk_size_mb:.4f} MB")
```
